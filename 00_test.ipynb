```python
# if elif else statement for marks of the students (range-based)
# above 90 grade A, 80-90 grade B, 70-80 grade C, 60-70 grade D, below 60 grade F
marks = int(input("Enter your marks: "))
if marks > 100 :
    print("Invalid marks. Please enter a value between 0 and 100.")
elif marks > 90:
    print("Grade: A")
elif marks >= 80:    
    print("Grade: B")
elif marks >= 70:
    print("Grade: C")
elif marks >= 60:  
    print("Grade: D")
else:
    print("Grade: F")
    
```
