```python
# Question 3,Part A
try:
    N = int(input("Enter the total number of candidate additives (N): "))
    R = int(input("Enter how many additives to test together (R): "))

    def factorial(n):
        if n < 0:
            print("Error: Factorial is not defined for negative integers.")
            return None
        if n == 0 or n == 1:
            return 1
        return n * factorial(n - 1)


    def combination(n, r):
        if n < 0 or r < 0 or r > n:
            print("Error: N and R must be non-negative integers, with R <= N.")
            return 0

        return factorial(n) // (factorial(r) * factorial(n - r))


    print(" The factorial of", N, "is", factorial(N))
    print("Number of possible formulations:", combination(N, R))

except ValueError as error:
    print("Invalid input:", error)
```

```python
# Question 3, Part B
def fib_iter(n):
    """Return a list of the first n Fibonacci numbers."""
    if n < 0:
        raise ValueError("n must be non-negative.")

    numbers = []
    a, b = 0, 1

    for _ in range(n):
        numbers.append(a)
        a, b = b, a + b

    return numbers


def fib_rec(k):
    """Return the k-th Fibonacci number, with F(0) = 0 and F(1) = 1."""
    if k < 0:
        raise ValueError("k must be non-negative.")
    if k <= 1:
        return k
    return fib_rec(k - 1) + fib_rec(k - 2)


choice = input("Choose method for Fibonacci sequence (1 , 2): ").strip().lower()
if choice == "1"  :
    n = int(input("How many Fibonacci numbers should be generated? "))
    fibonacci_numbers = fib_iter(n)
    print("Fibonacci numbers:", fibonacci_numbers)
elif choice == "2":
    k = int(input("Which Fibonacci index should be calculated recursively? "))
    print(f"F({k}) =", fib_rec(k))
else:
    print("Invalid choice.")
```
